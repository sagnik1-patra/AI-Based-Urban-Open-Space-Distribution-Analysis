```python

# ============================================================
# PARTICLE SWARM OPTIMIZATION (PSO)
# AI-BASED URBAN OPEN SPACE DISTRIBUTION ANALYSIS
# AND GREEN INFRASTRUCTURE OPTIMIZATION
#
# ALL OUTPUTS USE THE pso_ PREFIX
#
# Outputs:
# pso_accuracy_graph.png
# pso_heatmap.png
# pso_comparison_graph.png
# pso_result.csv
# pso_result_graph.png
# pso_prediction.csv
# pso_prediction_graph.png
# pso_convergence.csv
# pso_convergence_graph.png
# pso_priority_graph.png
# pso_model_metrics.csv
# pso_model.pkl
# pso_model.h5
# pso_config.yaml
# pso_results.json
# ============================================================

# STEP 1: INSTALL REQUIRED PACKAGES
import sys
import subprocess
import importlib.util

packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
    "h5py": "h5py",
    "yaml": "PyYAML"
}

for module, package in packages.items():
    if importlib.util.find_spec(module) is None:
        subprocess.check_call([
            sys.executable, "-m", "pip",
            "install", package
        ])

# ============================================================
# STEP 2: IMPORT LIBRARIES
# ============================================================

import json
import pickle
import warnings
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import h5py
import yaml

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

# ============================================================
# STEP 3: PROJECT PATHS
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Urban Open Space Distribution Analysis and Green Infrastructure Optimization Using PSO and AIS"
)

DATASET_PATH = (
    PROJECT_DIR / "OpenSpaces_Jalandhar_2.csv"
)

PROJECT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

print("=" * 75)
print("PSO-BASED URBAN OPEN SPACE OPTIMIZATION")
print("=" * 75)

print("\nDataset:", DATASET_PATH)
print("Output Directory:", PROJECT_DIR)

# ============================================================
# STEP 4: PSO HYPERPARAMETERS
# ============================================================

SEED = 42

# Match the previous AIS simulated budget.
ADDITIONAL_SPACES = 100

SWARM_SIZE = 50
ITERATIONS = 200

INERTIA_START = 0.9
INERTIA_END = 0.4

COGNITIVE_COEFFICIENT = 1.5
SOCIAL_COEFFICIENT = 1.5

VELOCITY_LIMIT = 20.0

# None means no externally verified capacity limit.
MAX_NEW_SPACES_PER_SECTOR = None

rng = np.random.default_rng(SEED)

print("\nPSO Configuration:")
print("Particles:", SWARM_SIZE)
print("Iterations:", ITERATIONS)
print("Simulated Budget:", ADDITIONAL_SPACES)

# ============================================================
# STEP 5: LOAD AND CLEAN DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

required_columns = [
    "City Name",
    "Zone Name",
    "Open space name/no"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

data = df[required_columns].copy()

data = data.rename(columns={
    "City Name": "city",
    "Zone Name": "sector",
    "Open space name/no": "open_space_count"
})

data["city"] = (
    data["city"].astype("string").str.strip()
)

data["sector"] = (
    data["sector"].astype("string").str.strip()
)

data["open_space_count"] = pd.to_numeric(
    data["open_space_count"],
    errors="coerce"
)

data = data.dropna(
    subset=["city", "sector", "open_space_count"]
).copy()

data = data[
    (data["city"] != "") &
    (data["sector"] != "") &
    (data["open_space_count"] >= 0)
].copy()

data["sector_number"] = pd.to_numeric(
    data["sector"].str.extract(
        r"(\d+)$", expand=False
    ),
    errors="coerce"
)

data = data.dropna(
    subset=["sector_number"]
).copy()

if not np.allclose(
    data["open_space_count"],
    np.round(data["open_space_count"])
):
    raise ValueError(
        "Open-space values are not integer counts."
    )

data["open_space_count"] = (
    data["open_space_count"].astype(int)
)

data["sector_number"] = (
    data["sector_number"].astype(int)
)

data = (
    data.groupby(
        ["city", "sector", "sector_number"],
        as_index=False
    )["open_space_count"]
    .sum()
)

data = data.sort_values(
    ["city", "sector_number"]
).reset_index(drop=True)

if data["sector_number"].duplicated().any():
    raise ValueError(
        "Duplicate sector numbers detected."
    )

if len(data) < 2:
    raise ValueError(
        "At least two sectors are required."
    )

existing = data[
    "open_space_count"
].to_numpy(dtype=np.int64)

sectors = data["sector"].tolist()

N = len(existing)
BUDGET = int(ADDITIONAL_SPACES)

if BUDGET < 0:
    raise ValueError("Budget must be nonnegative.")

print("\nDataset Shape:", df.shape)
print("Valid Sectors:", N)
print("Existing Open Spaces:", int(existing.sum()))

print("\nCleaned Dataset:")
print(data.to_string(index=False))

# ============================================================
# STEP 6: DEFINE CAPACITY CONSTRAINTS
# ============================================================

if MAX_NEW_SPACES_PER_SECTOR is None:
    capacities = np.full(
        N,
        BUDGET,
        dtype=np.int64
    )
else:
    capacities = np.full(
        N,
        int(MAX_NEW_SPACES_PER_SECTOR),
        dtype=np.int64
    )

if np.any(capacities < 0):
    raise ValueError(
        "Sector capacities must be nonnegative."
    )

if int(capacities.sum()) < BUDGET:
    raise ValueError(
        "Total capacity is smaller than the budget."
    )

# ============================================================
# STEP 7: DEFINE OBJECTIVE FUNCTIONS
# ============================================================

# Minimize variance in final open-space counts.
#
# x[i] = additional spaces assigned to sector i
#
# Constraints:
# x[i] >= 0
# x[i] <= capacity[i]
# sum(x) = BUDGET
#
# Final counts = existing + x
#
# All sectors have equal weight in this simulation.

def objective(allocation):
    final_counts = existing + allocation
    return float(np.var(final_counts))


def gini_coefficient(values):
    values = np.asarray(
        values,
        dtype=float
    )

    if values.sum() <= 0:
        return 0.0

    values = np.sort(values)

    n = len(values)

    return float(
        (
            2 * np.sum(
                np.arange(1, n + 1) * values
            ) / (n * values.sum())
        ) - (n + 1) / n
    )


def evaluate_allocation(allocation):
    final = existing + allocation

    return {
        "variance": float(np.var(final)),
        "standard_deviation": float(np.std(final)),
        "gini": gini_coefficient(final),
        "minimum_count": int(final.min()),
        "maximum_count": int(final.max()),
        "total_spaces": int(final.sum())
    }


def validate_allocation(x):
    x = np.asarray(x)

    return bool(
        len(x) == N
        and np.all(x >= 0)
        and np.all(x <= capacities)
        and np.allclose(x, np.round(x))
        and int(np.sum(x)) == BUDGET
    )

# ============================================================
# STEP 8: REPAIR CONTINUOUS PSO POSITIONS
# ============================================================

# PSO uses continuous position vectors.
# Convert them into feasible integer allocations.
#
# The repair method first clips and floors values,
# then adds/removes units until the budget is satisfied.

def repair_allocation(position):

    position = np.asarray(
        position,
        dtype=float
    )

    position = np.nan_to_num(
        position,
        nan=0.0,
        posinf=float(BUDGET),
        neginf=0.0
    )

    position = np.clip(
        position,
        0,
        capacities.astype(float)
    )

    if BUDGET == 0:
        return np.zeros(N, dtype=np.int64)

    if position.sum() <= 0:
        position = np.ones(N, dtype=float)

    # Scale the continuous position toward the budget.
    scaled = position * (
        BUDGET / position.sum()
    )

    scaled = np.minimum(
        scaled,
        capacities.astype(float)
    )

    allocation = np.floor(
        scaled
    ).astype(np.int64)

    remaining = BUDGET - int(
        allocation.sum()
    )

    # Add units to sectors with the largest
    # fractional remainders and available capacity.
    fractional = scaled - np.floor(scaled)

    while remaining > 0:

        eligible = np.where(
            allocation < capacities
        )[0]

        if len(eligible) == 0:
            raise RuntimeError(
                "Allocation repair failed."
            )

        # Prioritize fractional remainder.
        # Tie-break using lower resulting count.
        idx = max(
            eligible,
            key=lambda i: (
                fractional[i],
                -(existing[i] + allocation[i]),
                -i
            )
        )

        allocation[idx] += 1
        remaining -= 1

        # Once a fractional remainder has received
        # its unit, reduce its preference.
        fractional[idx] -= 1.0

    # Defensive handling of excess allocation.
    while remaining < 0:

        eligible = np.where(
            allocation > 0
        )[0]

        idx = max(
            eligible,
            key=lambda i: (
                allocation[i],
                existing[i] + allocation[i]
            )
        )

        allocation[idx] -= 1
        remaining += 1

    if not validate_allocation(allocation):
        raise RuntimeError(
            "Invalid repaired PSO allocation."
        )

    return allocation

# ============================================================
# STEP 9: INITIALIZE PSO PARTICLES
# ============================================================

positions = np.zeros(
    (SWARM_SIZE, N),
    dtype=float
)

velocities = rng.uniform(
    -1,
    1,
    size=(SWARM_SIZE, N)
)

for i in range(SWARM_SIZE):

    # Dirichlet initialization distributes the budget.
    weights = rng.dirichlet(
        np.ones(N)
    )

    positions[i] = (
        weights * BUDGET
    )

# Personal best positions and allocations.
pbest_positions = positions.copy()

pbest_allocations = np.array([
    repair_allocation(p)
    for p in positions
])

pbest_scores = np.array([
    objective(a)
    for a in pbest_allocations
])

best_idx = int(
    np.argmin(pbest_scores)
)

gbest_position = (
    pbest_positions[best_idx].copy()
)

gbest_allocation = (
    pbest_allocations[best_idx].copy()
)

gbest_score = float(
    pbest_scores[best_idx]
)

# ============================================================
# STEP 10: RUN PSO OPTIMIZATION
# ============================================================

best_history = []
mean_history = []

print("\n" + "=" * 75)
print("RUNNING PARTICLE SWARM OPTIMIZATION")
print("=" * 75)

for iteration in range(ITERATIONS):

    # Linearly decreasing inertia.
    if ITERATIONS > 1:
        progress = iteration / (ITERATIONS - 1)
    else:
        progress = 1.0

    inertia = (
        INERTIA_START
        - (INERTIA_START - INERTIA_END) * progress
    )

    r1 = rng.random(
        size=(SWARM_SIZE, N)
    )

    r2 = rng.random(
        size=(SWARM_SIZE, N)
    )

    # PSO velocity update.
    velocities = (
        inertia * velocities
        + COGNITIVE_COEFFICIENT
        * r1
        * (pbest_positions - positions)
        + SOCIAL_COEFFICIENT
        * r2
        * (gbest_position - positions)
    )

    velocities = np.clip(
        velocities,
        -VELOCITY_LIMIT,
        VELOCITY_LIMIT
    )

    # Position update.
    positions = positions + velocities

    positions = np.clip(
        positions,
        0,
        capacities.astype(float)
    )

    current_scores = []

    for i in range(SWARM_SIZE):

        allocation = repair_allocation(
            positions[i]
        )

        score = objective(allocation)

        current_scores.append(score)

        # Update personal best.
        if score < pbest_scores[i]:

            pbest_scores[i] = score

            pbest_positions[i] = (
                positions[i].copy()
            )

            pbest_allocations[i] = (
                allocation.copy()
            )

        # Update global best.
        if score < gbest_score:

            gbest_score = float(score)

            gbest_position = (
                positions[i].copy()
            )

            gbest_allocation = (
                allocation.copy()
            )

    best_history.append(gbest_score)

    mean_history.append(
        float(np.mean(current_scores))
    )

    if (
        (iteration + 1) % 20 == 0
        or iteration == 0
    ):
        print(
            f"Iteration {iteration + 1:3d} | "
            f"Best Variance: {gbest_score:.6f}"
        )

pso_solution = gbest_allocation.copy()
pso_score = float(gbest_score)

assert validate_allocation(pso_solution)

print("\nPSO Optimization Completed!")

# ============================================================
# STEP 11: EXACT GREEDY BENCHMARK
# ============================================================

# For the equal-weight variance objective, allocating
# each unit to the currently smallest eligible sector
# is an exact optimum.
#
# This provides a correctness reference for PSO.

def greedy_allocation():

    allocation = np.zeros(
        N,
        dtype=np.int64
    )

    for _ in range(BUDGET):

        eligible = np.where(
            allocation < capacities
        )[0]

        final = existing + allocation

        idx = eligible[
            np.argmin(final[eligible])
        ]

        allocation[idx] += 1

    return allocation


greedy_solution = greedy_allocation()
greedy_score = objective(greedy_solution)

if pso_score < greedy_score - 1e-8:
    raise AssertionError(
        "PSO outperformed the exact benchmark. "
        "Check the implementation."
    )

# ============================================================
# STEP 12: EVALUATE PERFORMANCE
# ============================================================

baseline_solution = np.zeros(
    N,
    dtype=np.int64
)

baseline_metrics = evaluate_allocation(
    baseline_solution
)

greedy_metrics = evaluate_allocation(
    greedy_solution
)

pso_metrics = evaluate_allocation(
    pso_solution
)

baseline_variance = baseline_metrics["variance"]

variance_reduction = (
    100 * (baseline_variance - pso_score)
    / baseline_variance
    if baseline_variance > 0
    else 0.0
)

optimality_gap = (
    100 * (pso_score - greedy_score)
    / greedy_score
    if greedy_score > 0
    else 0.0
)

print("\n" + "=" * 75)
print("PSO PERFORMANCE SUMMARY")
print("=" * 75)

print("Original Variance:", baseline_variance)
print("Greedy Variance:", greedy_score)
print("PSO Variance:", pso_score)

print(
    f"Variance Reduction: {variance_reduction:.2f}%"
)

print(
    f"Optimality Gap: {optimality_gap:.2f}%"
)

# ============================================================
# STEP 13: CREATE RESULT CSV
# ============================================================

result_df = data[
    ["city", "sector", "sector_number"]
].copy()

result_df["Existing_Open_Spaces"] = existing

result_df["PSO_Allocated_Spaces"] = (
    pso_solution
)

result_df["PSO_Final_Open_Spaces"] = (
    existing + pso_solution
)

result_df["Greedy_Allocated_Spaces"] = (
    greedy_solution
)

result_df["Greedy_Final_Open_Spaces"] = (
    existing + greedy_solution
)

result_df["PSO_Allocation_Share_Percent"] = (
    100 * pso_solution / BUDGET
    if BUDGET > 0
    else np.zeros(N)
)

result_df["PSO_Change_Percent"] = np.where(
    existing > 0,
    100 * pso_solution / np.maximum(existing, 1),
    np.nan
)

result_df.to_csv(
    PROJECT_DIR / "pso_result.csv",
    index=False
)

print("\nPSO RESULT CSV:")
print(result_df.to_string(index=False))

# ============================================================
# STEP 14: CREATE PREDICTION CSV
# ============================================================

# This is a simulated post-allocation outcome,
# not a statistical forecast.

prediction_df = result_df[
    [
        "city",
        "sector",
        "sector_number",
        "Existing_Open_Spaces",
        "PSO_Allocated_Spaces",
        "PSO_Final_Open_Spaces"
    ]
].copy()

prediction_df = prediction_df.rename(columns={
    "PSO_Final_Open_Spaces":
        "Predicted_Post_Allocation_Count"
})

prediction_df["Prediction_Type"] = (
    "Simulated allocation outcome"
)

prediction_df["Scenario_Budget"] = BUDGET

prediction_df.to_csv(
    PROJECT_DIR / "pso_prediction.csv",
    index=False
)

print("\nPSO PREDICTION CSV:")
print(prediction_df.to_string(index=False))

# ============================================================
# STEP 15: CREATE MODEL METRICS CSV
# ============================================================

comparison_df = pd.DataFrame([
    {
        "Method": "Original",
        **baseline_metrics
    },
    {
        "Method": "Greedy Exact Benchmark",
        **greedy_metrics
    },
    {
        "Method": "PSO",
        **pso_metrics
    }
])

comparison_df.to_csv(
    PROJECT_DIR / "pso_model_metrics.csv",
    index=False
)

print("\nPSO MODEL COMPARISON:")
print(comparison_df.to_string(index=False))

# ============================================================
# STEP 16: CREATE CONVERGENCE CSV
# ============================================================

convergence_df = pd.DataFrame({
    "Iteration": np.arange(
        1, ITERATIONS + 1
    ),
    "Best_Fitness": best_history,
    "Mean_Population_Fitness": mean_history,
    "Exact_Benchmark": np.full(
        ITERATIONS,
        greedy_score
    )
})

convergence_df.to_csv(
    PROJECT_DIR / "pso_convergence.csv",
    index=False
)

# ============================================================
# STEP 17: ACCURACY GRAPH
# ============================================================

# Optimization fitness is used instead of
# classification accuracy.
# Lower variance indicates better count balance.

plt.figure(figsize=(12, 6))

plt.plot(
    convergence_df["Iteration"],
    convergence_df["Best_Fitness"],
    linewidth=2,
    label="PSO Best Fitness"
)

plt.plot(
    convergence_df["Iteration"],
    convergence_df["Mean_Population_Fitness"],
    linewidth=1.5,
    alpha=0.7,
    label="Mean Swarm Fitness"
)

plt.axhline(
    greedy_score,
    linestyle="--",
    color="red",
    label="Exact Benchmark"
)

plt.title(
    "PSO Optimization Accuracy - Fitness Convergence",
    fontsize=15
)

plt.xlabel("Iteration")
plt.ylabel("Variance Objective (Lower Is Better)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 18: HEATMAP
# ============================================================

heatmap_data = result_df.set_index("sector")[
    [
        "Existing_Open_Spaces",
        "PSO_Allocated_Spaces",
        "PSO_Final_Open_Spaces"
    ]
]

plt.figure(figsize=(10, 11))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    linewidths=0.5
)

plt.title(
    "PSO Urban Open Space Allocation Heatmap",
    fontsize=14
)

plt.xlabel("Open Space Indicators")
plt.ylabel("Sectors")
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 19: COMPARISON GRAPH
# ============================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 6)
)

sns.barplot(
    data=comparison_df,
    x="Method",
    y="variance",
    color="steelblue",
    ax=axes[0]
)

axes[0].set_title(
    "Variance Comparison"
)

axes[0].set_ylabel(
    "Variance (Lower Is Better)"
)

axes[0].tick_params(
    axis="x",
    rotation=15
)

sns.barplot(
    data=comparison_df,
    x="Method",
    y="gini",
    color="seagreen",
    ax=axes[1]
)

axes[1].set_title(
    "Gini Coefficient Comparison"
)

axes[1].set_ylabel(
    "Gini (Lower Is Better)"
)

axes[1].tick_params(
    axis="x",
    rotation=15
)

plt.suptitle(
    "PSO vs Baseline Allocation Strategies",
    fontsize=15
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 20: RESULT GRAPH
# ============================================================

positions = np.arange(N)
width = 0.38

plt.figure(figsize=(15, 7))

plt.bar(
    positions - width / 2,
    existing,
    width,
    label="Existing Open Spaces"
)

plt.bar(
    positions + width / 2,
    existing + pso_solution,
    width,
    label="After PSO Allocation"
)

plt.xticks(
    positions,
    sectors,
    rotation=45
)

plt.xlabel("Sector")
plt.ylabel("Number of Open Spaces")

plt.title(
    "Existing vs PSO-Optimized Open Space Counts",
    fontsize=15
)

plt.legend()
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 21: PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(15, 7))

plt.plot(
    positions,
    existing,
    marker="o",
    linewidth=2,
    label="Existing Counts"
)

plt.plot(
    positions,
    existing + pso_solution,
    marker="s",
    linestyle="--",
    linewidth=2,
    label="PSO Simulated Final Counts"
)

plt.xticks(
    positions,
    sectors,
    rotation=45
)

plt.xlabel("Sector")
plt.ylabel("Number of Open Spaces")

plt.title(
    "PSO Scenario-Based Open Space Predictions",
    fontsize=15
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 22: CONVERGENCE GRAPH
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    np.arange(1, ITERATIONS + 1),
    best_history,
    linewidth=2,
    label="PSO Best Objective"
)

plt.axhline(
    greedy_score,
    linestyle="--",
    color="red",
    label="Exact Benchmark"
)

plt.title(
    "PSO Convergence Across Iterations",
    fontsize=15
)

plt.xlabel("Iteration")
plt.ylabel("Variance Objective")

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_convergence_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 23: ALLOCATION PRIORITY GRAPH
# ============================================================

priority_df = result_df.sort_values(
    "PSO_Allocated_Spaces",
    ascending=False
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=priority_df,
    y="sector",
    x="PSO_Allocated_Spaces",
    color="mediumseagreen"
)

plt.title(
    "PSO Allocation Priority Across Sectors",
    fontsize=15
)

plt.xlabel("Additional Open Spaces Allocated")
plt.ylabel("Sector")
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "pso_priority_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 24: SAVE PSO MODEL AS PKL
# ============================================================

pso_model = {
    "algorithm": "Particle Swarm Optimization",
    "variant": "Discrete PSO with Integer Repair",
    "objective": "Minimize final sector count variance",
    "existing_counts": existing,
    "best_allocation": pso_solution,
    "best_score": pso_score,
    "greedy_allocation": greedy_solution,
    "greedy_score": greedy_score,
    "convergence": np.array(best_history),
    "mean_convergence": np.array(mean_history),
    "sector_names": sectors,
    "budget": BUDGET,
    "parameters": {
        "swarm_size": SWARM_SIZE,
        "iterations": ITERATIONS,
        "inertia_start": INERTIA_START,
        "inertia_end": INERTIA_END,
        "cognitive_coefficient":
            COGNITIVE_COEFFICIENT,
        "social_coefficient":
            SOCIAL_COEFFICIENT,
        "velocity_limit": VELOCITY_LIMIT,
        "seed": SEED
    },
    "limitations": (
        "Equal-sector count-balancing simulation. "
        "Population, land area, and real capacity "
        "constraints are unavailable."
    )
}

pkl_path = PROJECT_DIR / "pso_model.pkl"

with open(pkl_path, "wb") as file:
    pickle.dump(
        pso_model,
        file,
        protocol=pickle.HIGHEST_PROTOCOL
    )

print("\nPKL Saved:", pkl_path)

# ============================================================
# STEP 25: SAVE PSO MODEL AS H5
# ============================================================

h5_path = PROJECT_DIR / "pso_model.h5"

with h5py.File(h5_path, "w") as h5:

    h5.attrs["algorithm"] = (
        "Particle Swarm Optimization"
    )

    h5.attrs["variant"] = (
        "Discrete PSO with Integer Repair"
    )

    h5.attrs["objective"] = (
        "Minimize final count variance"
    )

    h5.attrs["budget"] = BUDGET
    h5.attrs["best_score"] = pso_score

    h5.create_dataset(
        "existing_counts",
        data=existing
    )

    h5.create_dataset(
        "best_allocation",
        data=pso_solution
    )

    h5.create_dataset(
        "final_counts",
        data=existing + pso_solution
    )

    h5.create_dataset(
        "convergence",
        data=np.asarray(best_history)
    )

    h5.create_dataset(
        "mean_convergence",
        data=np.asarray(mean_history)
    )

    h5.create_dataset(
        "greedy_allocation",
        data=greedy_solution
    )

    string_type = h5py.string_dtype("utf-8")

    h5.create_dataset(
        "sectors",
        data=sectors,
        dtype=string_type
    )

print("H5 Saved:", h5_path)

# ============================================================
# STEP 26: SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project": (
        "AI-Based Urban Open Space Distribution "
        "Analysis and Green Infrastructure Optimization"
    ),
    "algorithm": "Particle Swarm Optimization",
    "variant": "Discrete PSO with Integer Repair",
    "dataset": str(DATASET_PATH),
    "objective": "Minimize variance of final counts",
    "budget": BUDGET,
    "swarm_size": SWARM_SIZE,
    "iterations": ITERATIONS,
    "inertia_start": INERTIA_START,
    "inertia_end": INERTIA_END,
    "cognitive_coefficient": COGNITIVE_COEFFICIENT,
    "social_coefficient": SOCIAL_COEFFICIENT,
    "velocity_limit": VELOCITY_LIMIT,
    "max_new_spaces_per_sector":
        MAX_NEW_SPACES_PER_SECTOR,
    "random_seed": SEED,
    "assumptions": [
        "All sectors have equal weight",
        "Every additional open space has unit value",
        "Allocation budget is simulated",
        "No verified land-capacity data",
        "No population-adjusted accessibility modeling"
    ]
}

yaml_path = PROJECT_DIR / "pso_config.yaml"

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("YAML Saved:", yaml_path)

# ============================================================
# STEP 27: SAVE JSON RESULTS
# ============================================================

def json_converter(obj):
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    raise TypeError(
        f"Cannot serialize {type(obj)}"
    )

json_results = {
    "algorithm": "Particle Swarm Optimization",
    "variant": "Discrete PSO with Integer Repair",
    "generated_at": datetime.now(
        timezone.utc
    ).isoformat(),
    "dataset": DATASET_PATH.name,
    "number_of_sectors": N,
    "simulated_budget": BUDGET,
    "baseline_metrics": baseline_metrics,
    "greedy_metrics": greedy_metrics,
    "pso_metrics": pso_metrics,
    "variance_improvement_percent":
        variance_reduction,
    "optimality_gap_percent":
        optimality_gap,
    "sector_results": result_df.to_dict(
        orient="records"
    ),
    "convergence": best_history,
    "limitations": config["assumptions"]
}

json_path = PROJECT_DIR / "pso_results.json"

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        json_results,
        file,
        indent=4,
        default=json_converter
    )

print("JSON Saved:", json_path)

# ============================================================
# STEP 28: VERIFY ALL GENERATED FILES
# ============================================================

generated_files = [
    "pso_accuracy_graph.png",
    "pso_heatmap.png",
    "pso_comparison_graph.png",
    "pso_result.csv",
    "pso_result_graph.png",
    "pso_prediction.csv",
    "pso_prediction_graph.png",
    "pso_convergence.csv",
    "pso_convergence_graph.png",
    "pso_priority_graph.png",
    "pso_model_metrics.csv",
    "pso_model.pkl",
    "pso_model.h5",
    "pso_config.yaml",
    "pso_results.json"
]

print("\n" + "=" * 75)
print("VERIFYING PSO OUTPUT FILES")
print("=" * 75)

for filename in generated_files:

    path = PROJECT_DIR / filename

    if path.exists():

        size_kb = (
            path.stat().st_size / 1024
        )

        print(
            f"[SUCCESS] {filename} "
            f"({size_kb:.2f} KB)"
        )

    else:
        print(f"[MISSING] {filename}")

# ============================================================
# STEP 29: FINAL SUMMARY
# ============================================================

print("\n" + "=" * 75)
print("FINAL PSO OPTIMIZATION SUMMARY")
print("=" * 75)

print(f"Total Sectors: {N}")
print(f"Existing Spaces: {existing.sum()}")
print(f"Additional Spaces: {BUDGET}")

print(
    f"Final Spaces: "
    f"{(existing + pso_solution).sum()}"
)

print(
    f"\nOriginal Variance: "
    f"{baseline_variance:.6f}"
)

print(
    f"Greedy Variance: "
    f"{greedy_score:.6f}"
)

print(
    f"PSO Variance: "
    f"{pso_score:.6f}"
)

print(
    f"Variance Reduction: "
    f"{variance_reduction:.2f}%"
)

print(
    f"Optimality Gap: "
    f"{optimality_gap:.2f}%"
)

print("\nAll PSO outputs saved in:")
print(PROJECT_DIR)

print("\nPSO EXECUTION COMPLETED SUCCESSFULLY")
```
