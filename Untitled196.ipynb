```python
import sys
import subprocess
import importlib.util

packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
    "h5py": "h5py",
    "yaml": "PyYAML"
}

for module, package in packages.items():
    if importlib.util.find_spec(module) is None:
        subprocess.check_call([
            sys.executable, "-m", "pip", "install", package
        ])

# 2. IMPORTS
import json
import pickle
import copy
import warnings
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import h5py
import yaml

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

# ============================================================
# 3. PATH CONFIGURATION
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Urban Open Space Distribution Analysis and Green Infrastructure Optimization Using PSO and AIS"
)

DATASET_PATH = PROJECT_DIR / "OpenSpaces_Jalandhar_2.csv"

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

# ============================================================
# 4. AIS CONFIGURATION
# ============================================================

SEED = 42

# SIMULATED planning budget, not an observed budget.
ADDITIONAL_SPACES = 100

POPULATION_SIZE = 50
GENERATIONS = 200
ELITE_COUNT = 10
CLONES_PER_ELITE = 8
RANDOM_INJECTION = 8
MAX_MUTATION_TRANSFERS = 10

# Set to None for no per-sector upper limit.
# The default assumes no land-capacity information.
MAX_NEW_SPACES_PER_SECTOR = None

# A fixed transfer can move several allocated units
# from one sector to another.
rng = np.random.default_rng(SEED)

print("=" * 72)
print("AIS-BASED URBAN OPEN SPACE OPTIMIZATION")
print("=" * 72)

# ============================================================
# 5. LOAD AND VALIDATE DATA
# ============================================================

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

required = [
    "City Name",
    "Zone Name",
    "Open space name/no"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(
        f"Missing dataset columns: {missing}"
    )

data = df[required].copy()

data = data.rename(columns={
    "City Name": "city",
    "Zone Name": "sector",
    "Open space name/no": "open_space_count"
})

data["city"] = data["city"].astype("string").str.strip()
data["sector"] = data["sector"].astype("string").str.strip()

data["open_space_count"] = pd.to_numeric(
    data["open_space_count"],
    errors="coerce"
)

data = data.dropna(
    subset=["city", "sector", "open_space_count"]
)

data = data[
    (data["city"] != "") &
    (data["sector"] != "") &
    (data["open_space_count"] >= 0)
].copy()

data["sector_number"] = pd.to_numeric(
    data["sector"].str.extract(
        r"(\d+)$", expand=False
    ),
    errors="coerce"
)

data = data.dropna(subset=["sector_number"])

if not np.allclose(
    data["open_space_count"],
    np.round(data["open_space_count"])
):
    raise ValueError(
        "Open-space values are not integer counts. "
        "Verify the source data."
    )

data["open_space_count"] = (
    data["open_space_count"].astype(int)
)

data["sector_number"] = (
    data["sector_number"].astype(int)
)

# Aggregate repeated sector rows, if present.
data = (
    data.groupby(
        ["city", "sector", "sector_number"],
        as_index=False
    )["open_space_count"]
    .sum()
)

data = data.sort_values(
    ["city", "sector_number"]
).reset_index(drop=True)

if data["sector_number"].duplicated().any():
    raise ValueError(
        "Duplicate sector numbers across cities. "
        "This implementation expects one city."
    )

if len(data) < 2:
    raise ValueError(
        "At least two sectors are required."
    )

sectors = data["sector"].tolist()

existing = data[
    "open_space_count"
].to_numpy(dtype=np.int64)

N = len(existing)
BUDGET = int(ADDITIONAL_SPACES)

if BUDGET < 0:
    raise ValueError("Budget cannot be negative.")

print("\nDataset shape:", df.shape)
print("Valid sectors:", N)
print("Existing listed spaces:", int(existing.sum()))
print("Simulated additional spaces:", BUDGET)

# ============================================================
# 6. OBJECTIVE FUNCTIONS
# ============================================================

# Objective: minimize variance of final counts.
#
# This assumes equal importance for every sector.
# It does not model population, land area, or access.
#
# The solution vector x has:
# x[i] >= 0
# sum(x) = BUDGET
#
# final_counts = existing + x

def variance_objective(allocation):
    final = existing + allocation
    return float(np.var(final))

def gini_coefficient(values):
    values = np.asarray(values, dtype=float)

    if np.sum(values) <= 0:
        return 0.0

    values = np.sort(values)
    n = len(values)

    return float(
        (2 * np.sum(
            np.arange(1, n + 1) * values
        ) / (n * np.sum(values)))
        - (n + 1) / n
    )

def get_metrics(allocation):
    final = existing + allocation

    return {
        "variance": float(np.var(final)),
        "standard_deviation": float(np.std(final)),
        "gini": gini_coefficient(final),
        "minimum_count": int(final.min()),
        "maximum_count": int(final.max()),
        "total_spaces": int(final.sum())
    }

# ============================================================
# 7. ALLOCATION CONSTRAINTS
# ============================================================

if MAX_NEW_SPACES_PER_SECTOR is None:
    capacities = np.full(
        N, BUDGET, dtype=np.int64
    )
else:
    capacities = np.full(
        N,
        int(MAX_NEW_SPACES_PER_SECTOR),
        dtype=np.int64
    )

if np.any(capacities < 0):
    raise ValueError("Capacities must be nonnegative.")

if capacities.sum() < BUDGET:
    raise ValueError(
        "Total sector capacity is below the budget."
    )

def validate_allocation(x):
    x = np.asarray(x)

    return (
        len(x) == N
        and np.all(x >= 0)
        and np.all(x <= capacities)
        and np.allclose(x, np.round(x))
        and int(x.sum()) == BUDGET
    )

# ============================================================
# 8. GENERATE FEASIBLE RANDOM SOLUTIONS
# ============================================================

def random_allocation():
    x = np.zeros(N, dtype=np.int64)

    remaining = BUDGET

    while remaining > 0:
        available = np.where(
            x < capacities
        )[0]

        if len(available) == 0:
            raise RuntimeError(
                "Unable to construct feasible allocation."
            )

        idx = int(rng.choice(available))
        x[idx] += 1
        remaining -= 1

    return x

# ============================================================
# 9. AIS HYPERMUTATION
# ============================================================

def mutate_allocation(solution, intensity=1):
    child = solution.copy()

    for _ in range(intensity):
        donors = np.where(child > 0)[0]

        receivers = np.where(
            child < capacities
        )[0]

        if len(donors) == 0 or len(receivers) == 0:
            break

        donor = int(rng.choice(donors))

        possible_receivers = receivers[
            receivers != donor
        ]

        if len(possible_receivers) == 0:
            break

        receiver = int(
            rng.choice(possible_receivers)
        )

        maximum_transfer = min(
            int(child[donor]),
            int(capacities[receiver] - child[receiver]),
            MAX_MUTATION_TRANSFERS
        )

        if maximum_transfer <= 0:
            continue

        amount = int(
            rng.integers(1, maximum_transfer + 1)
        )

        child[donor] -= amount
        child[receiver] += amount

    return child

# ============================================================
# 10. GREEDY BASELINE
# ============================================================

# Allocate each additional space to the sector
# with the smallest current count, subject to capacity.

def greedy_allocation():
    allocation = np.zeros(N, dtype=np.int64)

    for _ in range(BUDGET):
        eligible = np.where(
            allocation < capacities
        )[0]

        final = existing + allocation

        idx = eligible[
            np.argmin(final[eligible])
        ]

        allocation[idx] += 1

    return allocation

greedy_solution = greedy_allocation()
greedy_score = variance_objective(greedy_solution)

print("\nGreedy baseline variance:", greedy_score)

# ============================================================
# 11. AIS CLONAL SELECTION OPTIMIZER
# ============================================================

def run_ais():

    population = [
        random_allocation()
        for _ in range(POPULATION_SIZE)
    ]

    best_solution = None
    best_score = float("inf")

    history = []
    mean_history = []

    for generation in range(GENERATIONS):

        scores = np.array([
            variance_objective(x)
            for x in population
        ])

        order = np.argsort(scores)

        population = [
            population[i]
            for i in order
        ]

        scores = scores[order]

        if scores[0] < best_score:
            best_score = float(scores[0])
            best_solution = population[0].copy()

        # Clone the highest-affinity antibodies.
        selected = population[:ELITE_COUNT]

        clones = []

        for rank, antibody in enumerate(selected):

            # Higher-quality antibodies receive
            # smaller mutation intensity.
            intensity = max(
                1,
                int(
                    1 + rank * 4 /
                    max(1, ELITE_COUNT - 1)
                )
            )

            for _ in range(CLONES_PER_ELITE):
                clone = mutate_allocation(
                    antibody,
                    intensity=intensity
                )
                clones.append(clone)

        # Random antibody injection for diversity.
        newcomers = [
            random_allocation()
            for _ in range(RANDOM_INJECTION)
        ]

        candidates = (
            population +
            clones +
            newcomers +
            [best_solution.copy()]
        )

        # Remove duplicates while preserving
        # only feasible candidates.
        unique = {}

        for candidate in candidates:
            if validate_allocation(candidate):
                unique[tuple(candidate)] = candidate

        ranked = sorted(
            unique.values(),
            key=variance_objective
        )

        population = ranked[:POPULATION_SIZE]

        while len(population) < POPULATION_SIZE:
            population.append(random_allocation())

        generation_scores = [
            variance_objective(x)
            for x in population
        ]

        if min(generation_scores) < best_score:
            best_idx = int(
                np.argmin(generation_scores)
            )
            best_score = float(
                generation_scores[best_idx]
            )
            best_solution = (
                population[best_idx].copy()
            )

        history.append(best_score)
        mean_history.append(
            float(np.mean(generation_scores))
        )

        if (
            (generation + 1) % 20 == 0
            or generation == 0
        ):
            print(
                f"Generation {generation + 1:3d} | "
                f"Best variance: {best_score:.6f}"
            )

    return (
        best_solution,
        best_score,
        np.asarray(history),
        np.asarray(mean_history)
    )

print("\nRunning AIS optimization...")

ais_solution, ais_score, history, mean_history = (
    run_ais()
)

assert validate_allocation(ais_solution)

# ============================================================
# 12. EXACT EQUALIZATION BENCHMARK
# ============================================================

# With a single equal-weight count-balancing objective,
# the exact optimal allocation can be obtained greedily.
# This provides a correctness benchmark for AIS.

exact_solution = greedy_solution.copy()
exact_score = greedy_score

if ais_score < exact_score - 1e-8:
    raise AssertionError(
        "AIS unexpectedly outperformed the exact "
        "benchmark. Check the objective."
    )

# ============================================================
# 13. EVALUATE RESULTS
# ============================================================

zero_allocation = np.zeros(
    N, dtype=np.int64
)

baseline_metrics = get_metrics(zero_allocation)
greedy_metrics = get_metrics(greedy_solution)
ais_metrics = get_metrics(ais_solution)

baseline_variance = baseline_metrics["variance"]

improvement = (
    100 * (baseline_variance - ais_score)
    / baseline_variance
    if baseline_variance > 0
    else 0.0
)

optimality_gap = (
    100 * (ais_score - exact_score)
    / exact_score
    if exact_score > 0
    else 0.0
)

print("\n" + "=" * 72)
print("AIS OPTIMIZATION RESULTS")
print("=" * 72)

print("Original variance:", baseline_variance)
print("Greedy variance:", greedy_score)
print("AIS variance:", ais_score)
print("AIS variance reduction (%):", improvement)
print("AIS optimality gap (%):", optimality_gap)

# ============================================================
# 14. SAVE RESULT CSV
# ============================================================

result_df = data[
    ["city", "sector", "sector_number"]
].copy()

result_df["Existing_Open_Spaces"] = existing

result_df["AIS_Allocated_Spaces"] = ais_solution

result_df["AIS_Final_Open_Spaces"] = (
    existing + ais_solution
)

result_df["Greedy_Allocated_Spaces"] = (
    greedy_solution
)

result_df["Greedy_Final_Open_Spaces"] = (
    existing + greedy_solution
)

result_df["AIS_Allocation_Share_Percent"] = (
    100 * ais_solution / BUDGET
    if BUDGET > 0
    else np.zeros(N)
)

result_df["AIS_Change_Percent"] = np.where(
    existing > 0,
    100 * ais_solution / np.maximum(existing, 1),
    np.nan
)

result_df.to_csv(
    PROJECT_DIR / "ais_result.csv",
    index=False
)

print("\nAIS RESULT CSV:")
print(result_df.to_string(index=False))

# ============================================================
# 15. SAVE PREDICTION CSV
# ============================================================

# Prediction here means the scenario's resulting
# open-space counts after applying the AIS allocation.
# These are NOT statistical forecasts.

prediction_df = result_df[
    [
        "city",
        "sector",
        "sector_number",
        "Existing_Open_Spaces",
        "AIS_Allocated_Spaces",
        "AIS_Final_Open_Spaces"
    ]
].copy()

prediction_df = prediction_df.rename(columns={
    "AIS_Final_Open_Spaces":
        "Predicted_Post_Allocation_Count"
})

prediction_df["Prediction_Type"] = (
    "Simulated allocation outcome"
)

prediction_df["Scenario_Budget"] = BUDGET

prediction_df.to_csv(
    PROJECT_DIR / "ais_prediction.csv",
    index=False
)

# ============================================================
# 16. SAVE MODEL COMPARISON CSV
# ============================================================

comparison_df = pd.DataFrame([
    {
        "Method": "Original",
        **baseline_metrics
    },
    {
        "Method": "Greedy Exact Benchmark",
        **greedy_metrics
    },
    {
        "Method": "AIS",
        **ais_metrics
    }
])

comparison_df.to_csv(
    PROJECT_DIR / "ais_model_metrics.csv",
    index=False
)

print("\nMODEL COMPARISON:")
print(comparison_df.to_string(index=False))

# ============================================================
# 17. SAVE CONVERGENCE CSV
# ============================================================

convergence_df = pd.DataFrame({
    "Generation": np.arange(
        1, GENERATIONS + 1
    ),
    "Best_Fitness": history,
    "Mean_Population_Fitness": mean_history,
    "Exact_Benchmark": np.full(
        GENERATIONS, exact_score
    )
})

convergence_df.to_csv(
    PROJECT_DIR / "ais_convergence.csv",
    index=False
)

# ============================================================
# 18. ACCURACY GRAPH
# ============================================================

# This graph represents optimization fitness,
# NOT classification accuracy.
# Lower objective values are better.

plt.figure(figsize=(12, 6))

plt.plot(
    np.arange(1, GENERATIONS + 1),
    history,
    label="AIS Best Fitness",
    linewidth=2
)

plt.plot(
    np.arange(1, GENERATIONS + 1),
    mean_history,
    label="Mean Population Fitness",
    alpha=0.7
)

plt.axhline(
    exact_score,
    color="red",
    linestyle="--",
    label="Exact Benchmark"
)

plt.title(
    "AIS Optimization Accuracy - Objective Convergence"
)

plt.xlabel("Generation")
plt.ylabel("Variance Objective (Lower Is Better)")
plt.legend()
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 19. HEATMAP
# ============================================================

heatmap_data = result_df.set_index("sector")[
    [
        "Existing_Open_Spaces",
        "AIS_Allocated_Spaces",
        "AIS_Final_Open_Spaces"
    ]
]

plt.figure(figsize=(10, 11))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    linewidths=0.5
)

plt.title(
    "AIS Urban Open Space Allocation Heatmap"
)

plt.xlabel("Open Space Indicators")
plt.ylabel("Sectors")
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 20. COMPARISON GRAPH
# ============================================================

fig, axes = plt.subplots(
    1, 2, figsize=(14, 6)
)

sns.barplot(
    data=comparison_df,
    x="Method",
    y="variance",
    color="steelblue",
    ax=axes[0]
)

axes[0].set_title("Variance Comparison")
axes[0].set_ylabel("Variance (Lower Is Better)")
axes[0].tick_params(axis="x", rotation=15)

sns.barplot(
    data=comparison_df,
    x="Method",
    y="gini",
    color="seagreen",
    ax=axes[1]
)

axes[1].set_title("Gini Coefficient Comparison")
axes[1].set_ylabel("Gini (Lower Is Better)")
axes[1].tick_params(axis="x", rotation=15)

plt.suptitle(
    "AIS vs Baseline Allocation Strategies"
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 21. RESULT GRAPH
# ============================================================

positions = np.arange(N)
width = 0.38

plt.figure(figsize=(15, 7))

plt.bar(
    positions - width / 2,
    existing,
    width,
    label="Existing Spaces"
)

plt.bar(
    positions + width / 2,
    existing + ais_solution,
    width,
    label="After AIS Allocation"
)

plt.xticks(
    positions,
    sectors,
    rotation=45
)

plt.xlabel("Sector")
plt.ylabel("Number of Open Spaces")

plt.title(
    "Existing vs AIS-Optimized Open Space Counts"
)

plt.legend()
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 22. PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(15, 7))

plt.plot(
    positions,
    existing,
    marker="o",
    linewidth=2,
    label="Existing Count"
)

plt.plot(
    positions,
    existing + ais_solution,
    marker="s",
    linestyle="--",
    linewidth=2,
    label="AIS Simulated Final Count"
)

plt.xticks(
    positions,
    sectors,
    rotation=45
)

plt.xlabel("Sector")
plt.ylabel("Number of Open Spaces")

plt.title(
    "AIS Scenario-Based Open Space Predictions"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 23. CONVERGENCE GRAPH
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    convergence_df["Generation"],
    convergence_df["Best_Fitness"],
    linewidth=2,
    label="AIS Best Objective"
)

plt.axhline(
    exact_score,
    linestyle="--",
    color="red",
    label="Exact Benchmark"
)

plt.title("AIS Convergence Across Generations")
plt.xlabel("Generation")
plt.ylabel("Variance Objective")
plt.legend()
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_convergence_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 24. ALLOCATION PRIORITY GRAPH
# ============================================================

priority_df = result_df.sort_values(
    "AIS_Allocated_Spaces",
    ascending=False
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=priority_df,
    y="sector",
    x="AIS_Allocated_Spaces",
    color="mediumseagreen"
)

plt.title(
    "AIS Allocation Priority Across Sectors"
)

plt.xlabel("Additional Open Spaces Allocated")
plt.ylabel("Sector")
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "ais_priority_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 25. SAVE AIS MODEL AS PKL
# ============================================================

ais_model = {
    "algorithm": "Artificial Immune System",
    "variant": "Clonal Selection",
    "objective": "Minimize final sector count variance",
    "existing_counts": existing,
    "best_allocation": ais_solution,
    "best_score": ais_score,
    "greedy_allocation": greedy_solution,
    "greedy_score": greedy_score,
    "convergence": history,
    "mean_convergence": mean_history,
    "sector_names": sectors,
    "budget": BUDGET,
    "parameters": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "elite_count": ELITE_COUNT,
        "clones_per_elite": CLONES_PER_ELITE,
        "random_injection": RANDOM_INJECTION,
        "seed": SEED
    },
    "limitations": (
        "Equal-sector count-balancing simulation; "
        "no population, land-area or site-capacity evidence."
    )
}

with open(
    PROJECT_DIR / "ais_model.pkl",
    "wb"
) as file:
    pickle.dump(
        ais_model,
        file,
        protocol=pickle.HIGHEST_PROTOCOL
    )

# ============================================================
# 26. SAVE AIS MODEL AS H5
# ============================================================

with h5py.File(
    PROJECT_DIR / "ais_model.h5",
    "w"
) as h5:

    h5.attrs["algorithm"] = (
        "AIS Clonal Selection"
    )

    h5.attrs["objective"] = (
        "Minimize final count variance"
    )

    h5.attrs["budget"] = BUDGET
    h5.attrs["best_score"] = ais_score

    h5.create_dataset(
        "existing_counts",
        data=existing
    )

    h5.create_dataset(
        "best_allocation",
        data=ais_solution
    )

    h5.create_dataset(
        "final_counts",
        data=existing + ais_solution
    )

    h5.create_dataset(
        "convergence",
        data=history
    )

    h5.create_dataset(
        "mean_convergence",
        data=mean_history
    )

    h5.create_dataset(
        "greedy_allocation",
        data=greedy_solution
    )

    string_type = h5py.string_dtype("utf-8")

    h5.create_dataset(
        "sectors",
        data=sectors,
        dtype=string_type
    )

# ============================================================
# 27. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project": (
        "AI-Based Urban Open Space Distribution "
        "Analysis and Green Infrastructure Optimization"
    ),
    "algorithm": "AIS Clonal Selection",
    "dataset": str(DATASET_PATH),
    "objective": "Minimize variance of final counts",
    "budget": BUDGET,
    "population_size": POPULATION_SIZE,
    "generations": GENERATIONS,
    "elite_count": ELITE_COUNT,
    "clones_per_elite": CLONES_PER_ELITE,
    "random_injection": RANDOM_INJECTION,
    "max_mutation_transfers": MAX_MUTATION_TRANSFERS,
    "max_new_spaces_per_sector": (
        MAX_NEW_SPACES_PER_SECTOR
    ),
    "random_seed": SEED,
    "assumptions": [
        "All sectors have equal weight",
        "Each new open space has equal unit value",
        "All simulated allocations are feasible",
        "No real-world land constraints are modeled"
    ]
}

with open(
    PROJECT_DIR / "ais_config.yaml",
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

# ============================================================
# 28. SAVE JSON RESULTS
# ============================================================

def json_converter(obj):
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    raise TypeError(
        f"Cannot serialize {type(obj)}"
    )

json_results = {
    "algorithm": "AIS Clonal Selection",
    "generated_at": datetime.now(
        timezone.utc
    ).isoformat(),
    "dataset": DATASET_PATH.name,
    "number_of_sectors": N,
    "simulated_budget": BUDGET,
    "baseline_metrics": baseline_metrics,
    "greedy_metrics": greedy_metrics,
    "ais_metrics": ais_metrics,
    "variance_improvement_percent": improvement,
    "optimality_gap_percent": optimality_gap,
    "sector_results": result_df.to_dict(
        orient="records"
    ),
    "convergence": history.tolist(),
    "limitations": config["assumptions"]
}

with open(
    PROJECT_DIR / "ais_results.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        json_results,
        file,
        indent=4,
        default=json_converter
    )

# ============================================================
# 29. VERIFY GENERATED FILES
# ============================================================

generated_files = [
    "ais_accuracy_graph.png",
    "ais_heatmap.png",
    "ais_comparison_graph.png",
    "ais_result.csv",
    "ais_result_graph.png",
    "ais_prediction.csv",
    "ais_prediction_graph.png",
    "ais_convergence.csv",
    "ais_convergence_graph.png",
    "ais_priority_graph.png",
    "ais_model_metrics.csv",
    "ais_model.pkl",
    "ais_model.h5",
    "ais_config.yaml",
    "ais_results.json"
]

print("\n" + "=" * 72)
print("VERIFYING GENERATED AIS FILES")
print("=" * 72)

for filename in generated_files:
    path = PROJECT_DIR / filename

    if path.exists():
        size_kb = path.stat().st_size / 1024
        print(
            f"[SUCCESS] {filename} ({size_kb:.2f} KB)"
        )
    else:
        print(f"[MISSING] {filename}")

# ============================================================
# 30. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 72)
print("FINAL AIS OPTIMIZATION SUMMARY")
print("=" * 72)

print(f"Total sectors: {N}")
print(f"Existing spaces: {existing.sum()}")
print(f"Simulated additional spaces: {BUDGET}")
print(f"Final total spaces: {(existing + ais_solution).sum()}")

print(f"\nOriginal variance: {baseline_variance:.6f}")
print(f"Greedy variance: {greedy_score:.6f}")
print(f"AIS variance: {ais_score:.6f}")

print(
    f"AIS variance reduction: {improvement:.2f}%"
)

print(
    f"AIS optimality gap: {optimality_gap:.2f}%"
)

print("\nAll AIS outputs saved to:")
print(PROJECT_DIR)

print("\nAIS EXECUTION COMPLETED SUCCESSFULLY")
```
