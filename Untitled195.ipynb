```python
import sys
import subprocess
import importlib.util

packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
    "sklearn": "scikit-learn",
    "h5py": "h5py",
    "yaml": "PyYAML"
}

for module, package in packages.items():
    if importlib.util.find_spec(module) is None:
        subprocess.check_call([
            sys.executable, "-m", "pip",
            "install", package
        ])

# 2. IMPORT LIBRARIES
import json
import pickle
import warnings
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import h5py
import yaml

from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import LeaveOneOut, cross_val_predict
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")
np.random.seed(42)
sns.set_theme(style="whitegrid")

# ============================================================
# 3. SET PROJECT PATH
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Urban Open Space Distribution Analysis and Green Infrastructure Optimization Using PSO and AIS"
)

DATASET_PATH = PROJECT_DIR / "OpenSpaces_Jalandhar_2.csv"

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

print("=" * 70)
print("URBAN OPEN SPACE ANALYSIS AND ML EVALUATION")
print("=" * 70)

# ============================================================
# 4. LOAD AND CLEAN DATA
# ============================================================

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

print("\nOriginal dataset shape:", df.shape)
print("\nAvailable columns:")
print(df.columns.tolist())

required = [
    "City Name",
    "Zone Name",
    "Open space name/no"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(f"Missing columns: {missing}")

data = df[required].copy()

data.columns = [
    "city",
    "sector",
    "open_space_count"
]

data["city"] = data["city"].astype("string").str.strip()
data["sector"] = data["sector"].astype("string").str.strip()

data["open_space_count"] = pd.to_numeric(
    data["open_space_count"],
    errors="coerce"
)

data = data.dropna(
    subset=["city", "sector", "open_space_count"]
).copy()

data = data[
    (data["city"] != "") &
    (data["sector"] != "") &
    (data["open_space_count"] >= 0)
].copy()

# Extract actual sector number from strings like "Sector 1".
data["sector_number"] = pd.to_numeric(
    data["sector"].str.extract(
        r"(\d+)$", expand=False
    ),
    errors="coerce"
)

data = data.dropna(subset=["sector_number"])

data["sector_number"] = (
    data["sector_number"].astype(int)
)

if data["sector_number"].duplicated().any():
    raise ValueError(
        "Duplicate sector numbers found. "
        "Check whether records need aggregation."
    )

if len(data) < 5:
    raise ValueError(
        "At least 5 valid sectors are required."
    )

data = data.sort_values(
    "sector_number"
).reset_index(drop=True)

print("\nCleaned dataset:")
print(data.to_string(index=False))

# ============================================================
# 5. FEATURES AND TARGET
# ============================================================

# Only sector number is available as a numeric predictor.
# It is an identifier, not a proven causal predictor.
# This experiment measures whether any model can
# generalize from sector number to open-space count.

X = data[["sector_number"]]
y = data["open_space_count"].astype(float)

print("\nFeatures:", X.columns.tolist())
print("Target: open_space_count")
print("Number of records:", len(data))

# ============================================================
# 6. DEFINE REGRESSION MODELS
# ============================================================

models = {
    "Mean Baseline": DummyRegressor(
        strategy="mean"
    ),

    "Linear Regression": make_pipeline(
        StandardScaler(),
        LinearRegression()
    ),

    "Ridge Regression": make_pipeline(
        StandardScaler(),
        Ridge(alpha=10.0)
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        max_depth=3,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    )
}

# ============================================================
# 7. LEAVE-ONE-OUT CROSS-VALIDATION
# ============================================================

# With only 20 records, LOOCV gives each sector
# one out-of-fold prediction. These are not
# independent external test-set results.

cv = LeaveOneOut()

metrics_records = []
cv_predictions = {}

print("\n" + "=" * 70)
print("MODEL EVALUATION - LOOCV")
print("=" * 70)

for name, model in models.items():

    predictions = cross_val_predict(
        model,
        X,
        y,
        cv=cv,
        n_jobs=1
    )

    cv_predictions[name] = predictions

    mae = mean_absolute_error(y, predictions)

    mse = mean_squared_error(y, predictions)

    rmse = np.sqrt(mse)

    r2 = r2_score(y, predictions)

    metrics_records.append({
        "Model": name,
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2)
    })

    print(f"\n{name}")
    print(f"MAE  : {mae:.4f}")
    print(f"MSE  : {mse:.4f}")
    print(f"RMSE : {rmse:.4f}")
    print(f"R2   : {r2:.4f}")

metrics_df = pd.DataFrame(metrics_records)

metrics_df = metrics_df.sort_values(
    ["MAE", "RMSE"],
    ascending=True
).reset_index(drop=True)

metrics_df.to_csv(
    PROJECT_DIR / "model_metrics.csv",
    index=False
)

# Select best model using lowest out-of-fold MAE.
best_model_name = metrics_df.iloc[0]["Model"]

best_model = models[best_model_name]

best_cv_predictions = cv_predictions[
    best_model_name
]

# Fit the final model using all available data.
best_model.fit(X, y)

print("\nBest model by LOOCV MAE:", best_model_name)

# ============================================================
# 8. ACCURACY GRAPH (R2 SCORE)
# ============================================================

# R2 is a regression score, not classification accuracy.
# Negative R2 means worse than predicting the overall mean
# under the R2 evaluation definition.

plt.figure(figsize=(11, 6))

colors = [
    "seagreen" if v >= 0 else "indianred"
    for v in metrics_df["R2"]
]

bars = plt.bar(
    metrics_df["Model"],
    metrics_df["R2"],
    color=colors
)

plt.axhline(
    0,
    color="black",
    linewidth=1
)

for bar, value in zip(bars, metrics_df["R2"]):
    plt.annotate(
        f"{value:.3f}",
        (
            bar.get_x() + bar.get_width() / 2,
            value
        ),
        textcoords="offset points",
        xytext=(0, 5 if value >= 0 else -15),
        ha="center"
    )

plt.title(
    "Model Accuracy Comparison (LOOCV R2)",
    fontsize=15
)

plt.xlabel("Regression Models")
plt.ylabel("Cross-Validated R2 Score")
plt.xticks(rotation=15)
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 9. CORRELATION HEATMAP
# ============================================================

# Only actual observed numeric variables are included.
# Sector number is an identifier, so correlations
# must not be interpreted as causal relationships.

heatmap_data = data[
    ["sector_number", "open_space_count"]
].copy()

correlation = heatmap_data.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    square=True,
    linewidths=1
)

plt.title(
    "Correlation Heatmap - Urban Open Spaces",
    fontsize=14
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 10. MODEL COMPARISON GRAPH
# ============================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 6)
)

sns.barplot(
    data=metrics_df,
    x="Model",
    y="MAE",
    color="steelblue",
    ax=axes[0]
)

axes[0].set_title("Model Comparison - MAE")
axes[0].set_ylabel("Mean Absolute Error")
axes[0].tick_params(
    axis="x",
    rotation=25
)

sns.barplot(
    data=metrics_df,
    x="Model",
    y="RMSE",
    color="darkorange",
    ax=axes[1]
)

axes[1].set_title("Model Comparison - RMSE")
axes[1].set_ylabel("Root Mean Squared Error")
axes[1].tick_params(
    axis="x",
    rotation=25
)

plt.suptitle(
    "Cross-Validated Regression Model Comparison",
    fontsize=15
)

plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 11. RESULT CSV
# ============================================================

# Result CSV contains genuine out-of-fold predictions.
# Each sector was excluded when its prediction was made.

result_df = data[
    ["city", "sector", "sector_number", "open_space_count"]
].copy()

result_df = result_df.rename(columns={
    "open_space_count": "Actual"
})

result_df["Predicted"] = best_cv_predictions

result_df["Absolute_Error"] = np.abs(
    result_df["Actual"] -
    result_df["Predicted"]
)

result_df["Squared_Error"] = (
    result_df["Actual"] -
    result_df["Predicted"]
) ** 2

result_df["Evaluation"] = "LOOCV"

result_df.to_csv(
    PROJECT_DIR / "result.csv",
    index=False
)

print("\nRESULT CSV:")
print(result_df.to_string(index=False))

# ============================================================
# 12. RESULT GRAPH
# ============================================================

plt.figure(figsize=(14, 7))

plt.plot(
    result_df["sector_number"],
    result_df["Actual"],
    marker="o",
    linewidth=2,
    label="Actual Open Spaces"
)

plt.plot(
    result_df["sector_number"],
    result_df["Predicted"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="LOOCV Predicted Open Spaces"
)

plt.xlabel("Sector Number")
plt.ylabel("Number of Listed Open Spaces")

plt.title(
    f"Actual vs Out-of-Fold Predictions - {best_model_name}",
    fontsize=14
)

plt.xticks(
    result_df["sector_number"]
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 13. PREDICTION CSV
# ============================================================

# Generate fitted estimates for the same observed sectors.
# These are IN-SAMPLE predictions, not future forecasts.
# No new-sector values are fabricated.

fitted_predictions = best_model.predict(X)

prediction_df = data[
    ["city", "sector", "sector_number"]
].copy()

prediction_df["Fitted_Open_Space_Count"] = (
    fitted_predictions
)

prediction_df["Observed_Open_Space_Count"] = (
    y.to_numpy()
)

prediction_df["Prediction_Type"] = (
    "In-sample fitted estimate"
)

prediction_df.to_csv(
    PROJECT_DIR / "prediction.csv",
    index=False
)

print("\nPREDICTION CSV:")
print(prediction_df.to_string(index=False))

# ============================================================
# 14. PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(14, 7))

plt.scatter(
    prediction_df["sector_number"],
    prediction_df["Observed_Open_Space_Count"],
    label="Observed Counts",
    s=65
)

plt.plot(
    prediction_df["sector_number"],
    prediction_df["Fitted_Open_Space_Count"],
    marker="o",
    linestyle="--",
    linewidth=2,
    label="Fitted Model Estimates"
)

plt.title(
    f"In-Sample Open Space Estimates - {best_model_name}",
    fontsize=14
)

plt.xlabel("Sector Number")
plt.ylabel("Number of Listed Open Spaces")

plt.xticks(
    prediction_df["sector_number"]
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 15. ADDITIONAL RESIDUAL GRAPH
# ============================================================

residuals = (
    result_df["Actual"] -
    result_df["Predicted"]
)

plt.figure(figsize=(10, 6))

plt.scatter(
    result_df["Predicted"],
    residuals,
    s=65
)

plt.axhline(
    0,
    color="red",
    linestyle="--"
)

plt.xlabel("Out-of-Fold Predicted Count")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("Cross-Validated Residual Analysis")
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "residual_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 16. SAVE FINAL MODEL AS PKL
# ============================================================

model_artifact = {
    "model": best_model,
    "model_name": best_model_name,
    "features": ["sector_number"],
    "target": "open_space_count",
    "evaluation_method": "LeaveOneOut",
    "selection_metric": "MAE",
    "metrics": metrics_df.to_dict(orient="records"),
    "limitations": [
        "Only 20 observations",
        "Sector number is an identifier",
        "No future observations available",
        "No environmental covariates available",
        "Not a validated forecasting model"
    ]
}

pkl_path = PROJECT_DIR / "model.pkl"

with open(pkl_path, "wb") as file:
    pickle.dump(
        model_artifact,
        file,
        protocol=pickle.HIGHEST_PROTOCOL
    )

print("\nSaved:", pkl_path)

# ============================================================
# 17. SAVE MODEL DATA AS H5
# ============================================================

# HDF5 stores numerical model inputs, outputs and metrics.
# The fitted sklearn estimator itself is saved in model.pkl.

h5_path = PROJECT_DIR / "model.h5"

with h5py.File(h5_path, "w") as h5:

    h5.attrs["project"] = (
        "Urban Open Space Distribution Analysis"
    )

    h5.attrs["best_model"] = best_model_name
    h5.attrs["evaluation"] = "LOOCV"

    h5.create_dataset(
        "sector_number",
        data=X["sector_number"].to_numpy()
    )

    h5.create_dataset(
        "actual_counts",
        data=y.to_numpy()
    )

    h5.create_dataset(
        "cv_predictions",
        data=best_cv_predictions
    )

    h5.create_dataset(
        "fitted_predictions",
        data=fitted_predictions
    )

    h5.create_dataset(
        "residuals",
        data=residuals.to_numpy()
    )

    group = h5.create_group("model_metrics")

    for _, row in metrics_df.iterrows():
        safe_name = row["Model"].replace(" ", "_")
        subgroup = group.create_group(safe_name)

        for metric in ["MAE", "MSE", "RMSE", "R2"]:
            subgroup.attrs[metric] = float(row[metric])

print("Saved:", h5_path)

# ============================================================
# 18. SAVE YAML CONFIG
# ============================================================

config = {
    "project_name": (
        "AI-Based Urban Open Space Distribution Analysis "
        "and Green Infrastructure Optimization Using PSO and AIS"
    ),
    "dataset": str(DATASET_PATH),
    "random_state": 42,
    "target": "open_space_count",
    "features": ["sector_number"],
    "models": list(models.keys()),
    "evaluation": {
        "method": "Leave-One-Out Cross-Validation",
        "selection_metric": "MAE",
        "best_model": best_model_name
    },
    "outputs": {
        "accuracy_graph": "accuracy_graph.png",
        "heatmap": "heatmap.png",
        "comparison_graph": "comparison_graph.png",
        "result_csv": "result.csv",
        "result_graph": "result_graph.png",
        "prediction_csv": "prediction.csv",
        "prediction_graph": "prediction_graph.png",
        "metrics_csv": "model_metrics.csv"
    },
    "limitations": model_artifact["limitations"]
}

yaml_path = PROJECT_DIR / "config.yaml"

with open(yaml_path, "w", encoding="utf-8") as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("Saved:", yaml_path)

# ============================================================
# 19. SAVE JSON RESULTS
# ============================================================

def convert_json(value):
    if isinstance(value, np.integer):
        return int(value)
    if isinstance(value, np.floating):
        return float(value)
    if isinstance(value, np.ndarray):
        return value.tolist()
    raise TypeError(str(type(value)))

summary = {
    "project": config["project_name"],
    "generated_at": datetime.now(
        timezone.utc
    ).isoformat(),
    "total_records": len(data),
    "total_open_spaces": float(y.sum()),
    "best_model": best_model_name,
    "selection_metric": "LOOCV MAE",
    "model_comparison": metrics_df.to_dict(
        orient="records"
    ),
    "evaluation_results": result_df.to_dict(
        orient="records"
    ),
    "fitted_predictions": prediction_df.to_dict(
        orient="records"
    ),
    "limitations": model_artifact["limitations"]
}

json_path = PROJECT_DIR / "results.json"

with open(json_path, "w", encoding="utf-8") as file:
    json.dump(
        summary,
        file,
        indent=4,
        default=convert_json
    )

print("Saved:", json_path)

# ============================================================
# 20. VERIFY OUTPUT FILES
# ============================================================

files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "residual_graph.png",
    "model_metrics.csv",
    "model.pkl",
    "model.h5",
    "config.yaml",
    "results.json"
]

print("\n" + "=" * 70)
print("GENERATED FILES")
print("=" * 70)

for filename in files:
    path = PROJECT_DIR / filename

    if path.exists():
        print(
            f"[SUCCESS] {filename} "
            f"({path.stat().st_size / 1024:.2f} KB)"
        )
    else:
        print(f"[MISSING] {filename}")

print("\n" + "=" * 70)
print("FINAL MODEL SUMMARY")
print("=" * 70)

print("\nBest Model:", best_model_name)
print("\nCross-Validation Metrics:")
print(metrics_df.to_string(index=False))

print("\nResults saved at:")
print(PROJECT_DIR)

print("\nEXECUTION COMPLETED SUCCESSFULLY")
```
