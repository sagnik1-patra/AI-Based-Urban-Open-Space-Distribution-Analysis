```python
import sys
import subprocess
import importlib.util

required = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
    "h5py": "h5py",
    "yaml": "PyYAML"
}

for module, package in required.items():
    if importlib.util.find_spec(module) is None:
        subprocess.check_call([
            sys.executable, "-m", "pip", "install", package
        ])

# STEP 2: IMPORT LIBRARIES
import os
import json
import pickle
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import h5py
import yaml

import warnings
warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")
np.random.seed(42)

# ============================================================
# STEP 3: PROJECT PATHS
# ============================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Urban Open Space Distribution Analysis and Green Infrastructure Optimization Using PSO and AIS"
)

DATASET_PATH = PROJECT_DIR / "OpenSpaces_Jalandhar_2.csv"

PROJECT_DIR.mkdir(parents=True, exist_ok=True)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}\n"
        "Check the filename and folder location."
    )

print("=" * 65)
print("AI-BASED URBAN OPEN SPACE DISTRIBUTION ANALYSIS")
print("=" * 65)

print("\nProject Directory:", PROJECT_DIR)
print("Dataset Path:", DATASET_PATH)

# ============================================================
# STEP 4: LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

print("\nDataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst 5 Records:")
print(df.head().to_string(index=False))

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

# ============================================================
# STEP 5: IDENTIFY REQUIRED COLUMNS
# ============================================================

CITY_COLUMN = "City Name"
SECTOR_COLUMN = "Zone Name"
COUNT_COLUMN = "Open space name/no"

required_columns = [
    CITY_COLUMN,
    SECTOR_COLUMN,
    COUNT_COLUMN
]

missing_columns = [
    c for c in required_columns if c not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns missing: {missing_columns}"
    )

# Retain original data; do not fabricate missing
# ward, tree, area or typology measurements.

data = df[required_columns].copy()

data = data.rename(columns={
    CITY_COLUMN: "city",
    SECTOR_COLUMN: "sector",
    COUNT_COLUMN: "open_space_count"
})

data["city"] = data["city"].astype(str).str.strip()
data["sector"] = data["sector"].astype(str).str.strip()

data["open_space_count"] = pd.to_numeric(
    data["open_space_count"],
    errors="coerce"
)

data = data.dropna(
    subset=["city", "sector", "open_space_count"]
)

data = data[
    (data["city"] != "") &
    (data["sector"] != "") &
    (data["open_space_count"] >= 0)
].copy()

if data.empty:
    raise ValueError("No valid records remain.")

# Interpret numeric Open space name/no as a count.
# Confirm this interpretation with the data provider.

if not np.allclose(
    data["open_space_count"],
    np.round(data["open_space_count"])
):
    raise ValueError(
        "Open-space values are not integer counts. "
        "Check dataset interpretation."
    )

data["open_space_count"] = (
    data["open_space_count"].astype(int)
)

# Combine duplicate city-sector rows if present.
data = (
    data.groupby(
        ["city", "sector"],
        as_index=False
    )["open_space_count"]
    .sum()
)

# Sort sectors numerically.
data["sector_number"] = pd.to_numeric(
    data["sector"].str.extract(
        r"(\d+)$", expand=False
    ),
    errors="coerce"
)

data = data.sort_values(
    ["city", "sector_number", "sector"],
    na_position="last"
).reset_index(drop=True)

# ============================================================
# STEP 6: FEATURE ENGINEERING
# ============================================================

counts = data["open_space_count"].to_numpy(
    dtype=float
)

minimum = float(counts.min())
maximum = float(counts.max())
mean_count = float(counts.mean())
median_count = float(np.median(counts))
std_count = float(np.std(counts))

total_spaces = int(counts.sum())
total_sectors = int(len(data))

# Normalization based on observed counts.
denominator = maximum - minimum

if denominator > 0:
    normalized = (counts - minimum) / denominator
else:
    normalized = np.ones_like(counts)

data["normalized_open_spaces"] = normalized

# Relative priority:
# Fewer listed spaces -> higher exploratory priority.
# This is NOT a verified green-space deficit measure.
data["relative_priority_score"] = (
    100 * (1 - normalized)
)

# Descriptive comparison with observed mean.
data["difference_from_mean"] = (
    data["open_space_count"] - mean_count
)

data["open_space_share_percent"] = (
    100 * data["open_space_count"] / total_spaces
    if total_spaces > 0
    else 0.0
)

# Categorization using observed count quartiles.
q1 = float(np.percentile(counts, 25))
q3 = float(np.percentile(counts, 75))

def classify_sector(value):
    if np.isclose(minimum, maximum):
        return "Uniform Count"
    if value <= q1:
        return "Low Count"
    elif value >= q3:
        return "High Count"
    else:
        return "Moderate Count"

data["distribution_category"] = (
    data["open_space_count"].apply(classify_sector)
)

data["priority_rank"] = (
    data["relative_priority_score"]
    .rank(method="min", ascending=False)
    .astype(int)
)

# ============================================================
# STEP 7: DESCRIPTIVE STATISTICS
# ============================================================

# Gini coefficient measures concentration of counts,
# not equity of accessibility or land coverage.
def gini_coefficient(values):
    values = np.asarray(values, dtype=float)

    if len(values) == 0 or np.sum(values) == 0:
        return 0.0

    sorted_values = np.sort(values)
    n = len(sorted_values)

    return float(
        (2 * np.sum(
            np.arange(1, n + 1) * sorted_values
        ) / (n * np.sum(sorted_values)))
        - (n + 1) / n
    )

gini = gini_coefficient(counts)

statistics = {
    "total_sectors": total_sectors,
    "total_listed_open_spaces": total_spaces,
    "mean_open_spaces": mean_count,
    "median_open_spaces": median_count,
    "minimum_open_spaces": minimum,
    "maximum_open_spaces": maximum,
    "standard_deviation": std_count,
    "first_quartile": q1,
    "third_quartile": q3,
    "gini_coefficient": gini,
    "lowest_count_sector": str(
        data.loc[
            data["open_space_count"].idxmin(),
            "sector"
        ]
    ),
    "highest_count_sector": str(
        data.loc[
            data["open_space_count"].idxmax(),
            "sector"
        ]
    )
}

print("\n" + "=" * 65)
print("DATASET STATISTICS")
print("=" * 65)

for key, value in statistics.items():
    print(f"{key}: {value}")

# ============================================================
# STEP 8: SAVE PROCESSED DATA
# ============================================================

processed_path = PROJECT_DIR / "processed_data.csv"
sector_path = PROJECT_DIR / "sector_analysis.csv"

data.to_csv(processed_path, index=False)

sector_analysis = data.sort_values(
    ["priority_rank", "sector"]
).reset_index(drop=True)

sector_analysis.to_csv(
    sector_path,
    index=False
)

print("\nProcessed CSV files saved.")

# ============================================================
# STEP 9: SAVE PICKLE MODEL
# ============================================================

# This is a fitted statistical scoring model.
# No neural network or supervised predictor is trained.

model = {
    "model_name": "Urban Open Space Priority Scoring",
    "model_type": "Statistical Analysis",
    "version": "1.0",
    "features": ["open_space_count"],
    "minimum_count": minimum,
    "maximum_count": maximum,
    "mean_count": mean_count,
    "median_count": median_count,
    "first_quartile": q1,
    "third_quartile": q3,
    "priority_formula": (
        "100 * (1 - (count - min) / (max - min))"
    ),
    "statistics": statistics,
    "sector_results": sector_analysis.to_dict(
        orient="records"
    ),
    "note": (
        "Priority represents relative count scarcity only. "
        "It is not a measure of population-adjusted need."
    )
}

pkl_path = PROJECT_DIR / "model.pkl"

with open(pkl_path, "wb") as file:
    pickle.dump(
        model,
        file,
        protocol=pickle.HIGHEST_PROTOCOL
    )

print("PKL model saved:", pkl_path)

# ============================================================
# STEP 10: SAVE H5 FILE
# ============================================================

h5_path = PROJECT_DIR / "model.h5"

with h5py.File(h5_path, "w") as h5:
    h5.attrs["project_name"] = (
        "AI-Based Urban Open Space Distribution Analysis"
    )
    h5.attrs["model_type"] = (
        "Statistical Priority Scoring"
    )
    h5.attrs["created_at"] = (
        datetime.now(timezone.utc).isoformat()
    )

    h5.create_dataset(
        "open_space_counts",
        data=counts
    )

    h5.create_dataset(
        "normalized_open_spaces",
        data=data["normalized_open_spaces"].to_numpy()
    )

    h5.create_dataset(
        "priority_scores",
        data=data["relative_priority_score"].to_numpy()
    )

    h5.create_dataset(
        "priority_ranks",
        data=data["priority_rank"].to_numpy()
    )

    string_type = h5py.string_dtype(
        encoding="utf-8"
    )

    h5.create_dataset(
        "sector_names",
        data=data["sector"].tolist(),
        dtype=string_type
    )

    h5.create_dataset(
        "city_names",
        data=data["city"].tolist(),
        dtype=string_type
    )

    stats_group = h5.create_group("statistics")

    for key, value in statistics.items():
        if isinstance(value, str):
            stats_group.attrs[key] = value
        else:
            stats_group.attrs[key] = float(value)

print("H5 file saved:", h5_path)

# ============================================================
# STEP 11: SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project": {
        "name": (
            "AI-Based Urban Open Space Distribution Analysis "
            "and Green Infrastructure Optimization Using PSO and AIS"
        ),
        "version": "1.0",
        "city": "Jalandhar",
        "dataset": DATASET_PATH.name
    },
    "data": {
        "sector_column": SECTOR_COLUMN,
        "count_column": COUNT_COLUMN,
        "records": total_sectors,
        "missing_value_policy": (
            "Exclude unavailable measurements"
        )
    },
    "analysis": {
        "method": "Min-Max Priority Scoring",
        "normalization": "Min-Max",
        "higher_priority_means": (
            "Lower relative open-space count"
        ),
        "random_seed": 42
    },
    "optimization": {
        "pso": {
            "enabled": False,
            "particles": 30,
            "iterations": 100,
            "inertia_weight": 0.7,
            "cognitive_coefficient": 1.5,
            "social_coefficient": 1.5
        },
        "ais": {
            "enabled": False,
            "population_size": 30,
            "iterations": 100,
            "mutation_rate": 0.1
        }
    },
    "limitations": [
        "Only 20 sector-level observations",
        "Open-space area unavailable",
        "Tree counts unavailable",
        "Population unavailable",
        "Optimization requires explicit budget and constraints"
    ]
}

yaml_path = PROJECT_DIR / "config.yaml"

with open(yaml_path, "w", encoding="utf-8") as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("YAML configuration saved:", yaml_path)

# ============================================================
# STEP 12: SAVE JSON RESULTS
# ============================================================

def json_converter(obj):
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    if isinstance(obj, (datetime, pd.Timestamp)):
        return obj.isoformat()
    raise TypeError(
        f"Object of type {type(obj)} is not JSON serializable"
    )

results = {
    "project_name": config["project"]["name"],
    "generated_at": datetime.now(
        timezone.utc
    ).isoformat(),
    "dataset": DATASET_PATH.name,
    "statistics": statistics,
    "sector_results": sector_analysis.to_dict(
        orient="records"
    ),
    "methodology": {
        "normalization": "Min-Max",
        "priority_scoring": (
            "Inverse normalized open-space count"
        ),
        "gini": "Concentration of observed counts"
    },
    "limitations": config["limitations"]
}

json_path = PROJECT_DIR / "results.json"

with open(json_path, "w", encoding="utf-8") as file:
    json.dump(
        results,
        file,
        indent=4,
        ensure_ascii=False,
        default=json_converter
    )

print("JSON results saved:", json_path)

# ============================================================
# STEP 13: VISUALIZATION - OPEN SPACE DISTRIBUTION
# ============================================================

plt.figure(figsize=(14, 6))

sns.barplot(
    data=data,
    x="sector",
    y="open_space_count",
    color="steelblue"
)

plt.title(
    "Urban Open Space Distribution Across Jalandhar Sectors",
    fontsize=15
)
plt.xlabel("Sector")
plt.ylabel("Number of Listed Open Spaces")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "sector_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 14: VISUALIZATION - PRIORITY SCORES
# ============================================================

plt.figure(figsize=(14, 7))

priority_plot = sector_analysis.sort_values(
    "relative_priority_score",
    ascending=False
)

sns.barplot(
    data=priority_plot,
    x="sector",
    y="relative_priority_score",
    color="seagreen"
)

plt.title(
    "Relative Open Space Count Priority by Sector",
    fontsize=15
)
plt.xlabel("Sector")
plt.ylabel("Relative Priority Score (0-100)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "priority_scores.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 15: VISUALIZATION - HEATMAP
# ============================================================

heatmap_data = data.set_index("sector")[
    [
        "open_space_count",
        "normalized_open_spaces",
        "relative_priority_score"
    ]
]

# Standardize columns for visualization only.
heatmap_scaled = heatmap_data.copy()

for column in heatmap_scaled.columns:
    values = heatmap_scaled[column]
    spread = values.max() - values.min()

    if spread > 0:
        heatmap_scaled[column] = (
            (values - values.min()) / spread
        )
    else:
        heatmap_scaled[column] = 0.0

plt.figure(figsize=(9, 10))

sns.heatmap(
    heatmap_scaled,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    linewidths=0.5
)

plt.title(
    "Normalized Sector-Level Open Space Indicators"
)
plt.xlabel("Indicators")
plt.ylabel("Sectors")
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 16: VISUALIZATION - DISTRIBUTION CATEGORIES
# ============================================================

category_counts = (
    data["distribution_category"]
    .value_counts()
)

plt.figure(figsize=(8, 6))

plt.pie(
    category_counts.values,
    labels=category_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sector Distribution by Open Space Count")
plt.tight_layout()

plt.savefig(
    PROJECT_DIR / "distribution_categories.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 17: VERIFY SAVED FILES
# ============================================================

print("\n" + "=" * 65)
print("VERIFYING SAVED ARTIFACTS")
print("=" * 65)

with open(pkl_path, "rb") as file:
    loaded_model = pickle.load(file)

print(
    "PKL loaded:",
    loaded_model["model_name"]
)

with h5py.File(h5_path, "r") as file:
    print("H5 datasets:", list(file.keys()))
    print(
        "H5 record count:",
        len(file["open_space_counts"])
    )

with open(yaml_path, "r", encoding="utf-8") as file:
    loaded_config = yaml.safe_load(file)

print(
    "YAML loaded:",
    loaded_config["project"]["version"]
)

with open(json_path, "r", encoding="utf-8") as file:
    loaded_results = json.load(file)

print(
    "JSON loaded:",
    loaded_results["statistics"]["total_sectors"],
    "sectors"
)

# ============================================================
# STEP 18: DISPLAY FINAL RESULTS
# ============================================================

print("\n" + "=" * 65)
print("TOP SECTORS BY RELATIVE COUNT PRIORITY")
print("=" * 65)

print(
    sector_analysis[
        [
            "sector",
            "open_space_count",
            "relative_priority_score",
            "priority_rank",
            "distribution_category"
        ]
    ].head(10).to_string(index=False)
)

print("\n" + "=" * 65)
print("GENERATED FILES")
print("=" * 65)

generated_files = [
    "model.h5",
    "model.pkl",
    "config.yaml",
    "results.json",
    "processed_data.csv",
    "sector_analysis.csv",
    "sector_distribution.png",
    "priority_scores.png",
    "heatmap.png",
    "distribution_categories.png"
]

for filename in generated_files:
    filepath = PROJECT_DIR / filename

    if filepath.exists():
        size_kb = filepath.stat().st_size / 1024
        print(
            f"[SUCCESS] {filename} ({size_kb:.2f} KB)"
        )
    else:
        print(f"[MISSING] {filename}")

print("\n" + "=" * 65)
print("PROJECT EXECUTION COMPLETED SUCCESSFULLY")
print("=" * 65)

print("\nAll results saved in:")
print(PROJECT_DIR)
```
